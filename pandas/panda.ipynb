```python
import pandas as pd
```

# Dataframe from lists

```python
content = [1,3,5,7,9]
df = pd.DataFrame(content, columns=["Odd Number"])
print(df)
```

# Dataframe from list inside a list

```python
content = [[1,"Apple"],[2,"Banana"],[3,"Guava"],[4,"Kiwi"],[5,"Mango"]]
df = pd.DataFrame(content,columns=["Number","Fruits"])
print(df)
```

# DataFrame from Dictionaries

```python
content = [{"name": "Apple", "color": "Red"}, {"name": "Mango", "color": "Yellow"}, {"name": "Lychee", "color": "Pink"}]
df = pd.DataFrame(content)
print(df)
print()
content = [{"name": "Apple", "color": "Red","units":300}, {"name": "Mango", "color": "Yellow"}, {"name": "Lychee", "color": "Pink","units":400}]
df = pd.DataFrame(content)
print(df)
```

# DataFrame with Series

```python
content = {"One":pd.Series([1,2,3], index=["a","b","c"]),
      "Two": pd.Series([1,2,3,4], index=["a","b","c","d"])}
df = pd.DataFrame(content)
print(df)
print()
print(df["One"])
print()
print(df["Two"])

```
