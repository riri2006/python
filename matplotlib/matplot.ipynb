```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = [56,52,54,60,58,57,53]
b=["DL","CC","UDF","ICS","OE","BMM","HRM"]

salesProduct1 = [69,58,10,67,43,48,79,80]
salesProduct2=[80,79,32,56,78,69,87,89]
years=[2018,2019,2020,2021,2022,2023,2024,2025]
```

# Scatter Plot

```python
plt.scatter(b,x, color="green",marker="^")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

# Bar Graph

```python
plt.barh(b,x,color="green")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

# Line graph with numpy


```python
x=np.linspace(0,10*np.pi,100)
y_sin =np.sin(x)
y_cos=np.cos(x)
plt.subplot(2,2,1)
plt.plot(x, y_sin, color="purple")
plt.title("Sine")

plt.subplot(2,2,2)
plt.plot(x, y_cos, color="purple")
plt.title("Cosine")
```

# 2-Line graph

```python
plt.plot(years,salesProduct1, color="red")
plt.plot(years,salesProduct2)
plt.xlabel("Years")
plt.ylabel("Sales")

```
