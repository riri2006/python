```python
from matplotlib import pyplot as plt
```

```python
x = [56,52,54,60,58,57,53]
b=["DL","CC","UDF","ICS","OE","BMM","HRM"]
a=[42,44,56,35,60,39,45]
y=["DL","CC","UDF","ICS","OE","BMM","HRM"]
```

# Scatter Plot

```python
plt.scatter(y,x, color="green",marker="^")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

# Bar Graph

```python
plt.bar(y,x,color="green")
plt.bar(b,a)
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```
