```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = [56,52,54,60,58,57,53]
b=["DL","CC","UDF","ICS","OE","BMM","HRM"]
```

# Scatter Plot

```python
plt.scatter(b,x, color="green",marker="^")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

# Bar Graph

```python
plt.barh(b,x,color="green")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

# Line graph with numpy


```python
x=np.linspace(0,10*np.pi,100)
y_sin =np.sin(x)
y_cos=np.cos(x)
plt.subplot(2,2,1)
plt.plot(x, y_sin, color="purple")
plt.title("Sine")

plt.subplot(2,2,2)
plt.plot(x, y_cos, color="purple")
plt.title("Cosine")
```
